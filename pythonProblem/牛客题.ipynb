```python
#!/usr/bin/env python
# -*- coding: UTF-8 -*-
from collections import Counter

data = [
    "4",
    "9",
    "ababbcccc",
    "6",
    "dazzzd",
    "2",
    "ba",
    "7",
    "eeedddc"
]


def solve(n, s):
    strn = Counter(s)
    strn = sorted(strn.items(), key=lambda x: x[0])
    nums = [x[1] for x in strn]
    if len(nums) == 1:
        return "Yes"
    for i in range(len(nums) - 1):
        if nums[i] + 1 != nums[i + 1]:
            return "No"
    return "Yes"


def commit():
    n = int(data[0])  # 测试用例数量
    index = 1  # 从第二个元素开始
    for i in range(n):
        # 读取长度（虽然你的solve函数没用到，但需要读取来移动指针）
        length = int(data[index])
        index += 1
        # 读取字符串
        s = data[index]
        index += 1
        # 调用solve函数并输出结果
        result = solve(length, s)
        print(result)


commit()
```

```python
a = 3
b = [1,3,4]
# a = int(input())
# b = list(map(int, input().split()))
def solve(a,b):

    num = []
    if a == 0 or a == 1:
        return "Yes"
    for i in range(a-1):
        if b[i] >= b[i+1]:
            return "No"
        num.append(b[i+1]-b[i])
    print(num)
    if len(num) == 1 or len(num) == 0:
        return "Yes"
    for i in range(len(num)-1):
        if num[i] <= num[i+1]:
            return "No"
    return "Yes"

print(solve(a,b))
    
```

```python
num = 1
s = "12345aB."
def solve(s):
    dic = {"num":0, "little":0 ,"big":0 ,"other":0}
    dic2 = {"num":9, "little":25 ,"big":25 ,"other":3}
    total = 65
    sum = 0
    for i in s:
        if i.isdigit():
            dic["num"] += 1
        elif i.islower():
            dic["little"] += 1
        elif i.isupper():
            dic["big"] += 1
        else:
            dic["other"] += 1
    for k,v in dic.items():
        if v == 1:
            sum += dic2[k]
        else:
            sum += v*total
    return sum
print(solve(s))
```
